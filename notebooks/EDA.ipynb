```python
# ==========================================
# STEP 8: DETECT OUTLIERS USING IQR
# ==========================================

# Select numerical columns
numeric_columns = web.select_dtypes(
    include="number"
).columns

# Check outliers in each numerical column
for column in numeric_columns:

    Q1 = web[column].quantile(0.25)

    Q3 = web[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = web[
        (web[column] < lower_limit) |
        (web[column] > upper_limit)
    ]

    print(column, ":", len(outliers), "outliers")
```

```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# Load website traffic dataset

web = pd.read_csv("../dataset/website_wata.csv")

print("Dataset loaded successfully!")

print("Dataset Shape:", web.shape)
```

```python
# Display first five records

web.head()
```

```python

# ==========================================
# STEP 9: VISUALIZE OUTLIERS USING BOXPLOTS
# ==========================================

# Select numerical columns
numeric_columns = web.select_dtypes(
    include="number"
).columns

# Set figure size
plt.figure(figsize=(15, 10))

# Create boxplots
for i, column in enumerate(numeric_columns, 1):

    plt.subplot(2, 3, i)

    sns.boxplot(y=web[column])

    plt.title(column)

# Adjust layout
plt.tight_layout()

# Display charts
plt.show()
```

```python

# ==========================================
# STEP 10: CONVERSION RATE ANALYSIS
# ==========================================

# Display statistical summary
print("Conversion Rate Statistics:")

print(web["Conversion Rate"].describe())

# Count records with conversion rate = 1
print("\nConversion Rate = 1:")

print((web["Conversion Rate"] == 1).sum())

# Count unique conversion rate values
print("\nUnique Conversion Rate Values:")

print(web["Conversion Rate"].nunique())
```

```python

# ==========================================
# STEP 11: CONVERSION RATE HISTOGRAM
# ==========================================

# Set figure size
plt.figure(figsize=(10, 5))

# Create histogram
sns.histplot(
    data=web,
    x="Conversion Rate",
    bins=20,
    kde=True
)

# Add chart title
plt.title("Website Conversion Rate Distribution")

# Add axis labels
plt.xlabel("Conversion Rate")

plt.ylabel("Number of Records")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 12: CREATE CLEANED DATASET
# ==========================================

# Create a copy of the original dataset
cleaned_web = web.copy()

# Save cleaned data into CSV file
cleaned_web.to_csv(
    "../dataset/cleaned_website_traffic.csv",
    index=False
)

# Display confirmation
print("Cleaned Dataset Saved Successfully!")

print("Total Rows:", cleaned_web.shape[0])

print("Total Columns:", cleaned_web.shape[1])
```

```python
# ==========================================
# STEP 1: LOAD CLEANED DATASET
# ==========================================

# Load cleaned CSV file
cleaned_web = pd.read_csv(
    "../dataset/cleaned_website_traffic.csv"
)

# Display success message
print("Cleaned Dataset Loaded Successfully!")

# Display first five records
cleaned_web.head()
```

```python
# ==========================================
# STEP 2: TRAFFIC SOURCE ANALYSIS
# ==========================================

# Count visitors from each traffic source
traffic_counts = cleaned_web["Traffic Source"].value_counts()

# Display traffic source counts
print("Website Traffic Sources:")

print(traffic_counts)
```

```python
# ==========================================
# STEP 3: TRAFFIC SOURCE BAR CHART
# ==========================================

# Set chart size
plt.figure(figsize=(10, 5))

# Create bar chart
sns.countplot(
    data=cleaned_web,
    x="Traffic Source",
    order=cleaned_web["Traffic Source"].value_counts().index
)

# Add chart title
plt.title("Website Traffic by Source")

# Add axis labels
plt.xlabel("Traffic Source")
plt.ylabel("Number of Visitors")

# Display chart
plt.show()
```

```python
# ==========================================
# STEP 4: TRAFFIC SOURCE PERCENTAGE
# ==========================================

# Calculate percentage of each traffic source
traffic_percentage = (
    cleaned_web["Traffic Source"]
    .value_counts(normalize=True) * 100
)

# Display percentages
print("Traffic Source Percentage:")

print(traffic_percentage.round(2))
```

```python
# ==========================================
# STEP 5: TRAFFIC DISTRIBUTION PIE CHART
# ==========================================

# Calculate traffic source counts
traffic_counts = cleaned_web["Traffic Source"].value_counts()

# Set figure size
plt.figure(figsize=(8, 8))

# Create pie chart
plt.pie(
    traffic_counts,
    labels=traffic_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

# Add title
plt.title("Website Traffic Distribution")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 6: TRAFFIC SOURCE PERFORMANCE
# ==========================================

# Group data by traffic source
traffic_performance = cleaned_web.groupby(
    "Traffic Source"
).agg({
    "Page Views": "mean",
    "Session Duration": "mean",
    "Bounce Rate": "mean",
    "Time on Page": "mean",
    "Previous Visits": "mean",
    "Conversion Rate": "mean"
})

# Display results
print("Traffic Source Performance Analysis:")

display(traffic_performance.round(2))
```

```python

# ==========================================
# STEP 7: CONVERSION RATE COMPARISON
# ==========================================

# Calculate average conversion rate
conversion_data = cleaned_web.groupby(
    "Traffic Source"
)["Conversion Rate"].mean().sort_values(
    ascending=False
)

# Display results
print("Average Conversion Rate by Traffic Source:")

print(conversion_data.round(3))
```

```python

# ==========================================
# STEP 8: CONVERSION RATE BAR CHART
# ==========================================

# Set chart size
plt.figure(figsize=(10, 5))

# Create bar chart
sns.barplot(
    x=conversion_data.index,
    y=conversion_data.values
)

# Add title
plt.title("Average Conversion Rate by Traffic Source")

# Add labels
plt.xlabel("Traffic Source")
plt.ylabel("Average Conversion Rate")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 9: SESSION DURATION ANALYSIS
# ==========================================

# Calculate average session duration
session_data = cleaned_web.groupby(
    "Traffic Source"
)["Session Duration"].mean().sort_values(
    ascending=False
)

# Display results
print("Average Session Duration:")

print(session_data.round(2))

# Create bar chart
plt.figure(figsize=(10, 5))

sns.barplot(
    x=session_data.index,
    y=session_data.values
)

plt.title("Average Session Duration by Traffic Source")

plt.xlabel("Traffic Source")

plt.ylabel("Average Session Duration")

plt.show()
```

```python

# ==========================================
# STEP 10: BOUNCE RATE ANALYSIS
# ==========================================

# Calculate average bounce rate by traffic source

bounce_data = cleaned_web.groupby(
    "Traffic Source"
)["Bounce Rate"].mean().sort_values(
    ascending=False
)

# Display results

print("Average Bounce Rate by Traffic Source:")

print(bounce_data.round(3))
```

```python

# ==========================================
# STEP 11: BOUNCE RATE BAR CHART
# ==========================================

# Set figure size
plt.figure(figsize=(10, 5))

# Create bar chart
sns.barplot(
    x=bounce_data.index,
    y=bounce_data.values
)

# Add chart title
plt.title("Average Bounce Rate by Traffic Source")

# Add axis labels
plt.xlabel("Traffic Source")

plt.ylabel("Average Bounce Rate")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 12: PAGE VIEWS ANALYSIS
# ==========================================

# Calculate average page views
page_views_data = cleaned_web.groupby(
    "Traffic Source"
)["Page Views"].mean().sort_values(
    ascending=False
)

# Display results
print("Average Page Views by Traffic Source:")

print(page_views_data.round(2))

# Create bar chart
plt.figure(figsize=(10, 5))

sns.barplot(
    x=page_views_data.index,
    y=page_views_data.values
)

plt.title("Average Page Views by Traffic Source")

plt.xlabel("Traffic Source")

plt.ylabel("Average Page Views")

plt.show()
```

```python

# ==========================================
# STEP 13: TIME ON PAGE ANALYSIS
# ==========================================

# Calculate average time on page
time_data = cleaned_web.groupby(
    "Traffic Source"
)["Time on Page"].mean().sort_values(
    ascending=False
)

# Display results
print("Average Time on Page by Traffic Source:")

print(time_data.round(2))

# Create bar chart
plt.figure(figsize=(10, 5))

sns.barplot(
    x=time_data.index,
    y=time_data.values
)

plt.title("Average Time on Page by Traffic Source")

plt.xlabel("Traffic Source")

plt.ylabel("Average Time on Page")

plt.show()
```

```python

# ==========================================
# STEP 14: CONVERSION RATE ANALYSIS
# ==========================================

# Calculate average conversion rate by traffic source

conversion_data = cleaned_web.groupby(
    "Traffic Source"
)["Conversion Rate"].mean().sort_values(
    ascending=False
)

# Display results

print("Conversion Rate by Traffic Source:")

print(conversion_data.round(3))
```

```python

# ==========================================
# STEP 15: CONVERSION RATE VISUALIZATION
# ==========================================

# Set chart size
plt.figure(figsize=(10, 5))

# Create bar chart
sns.barplot(
    x=conversion_data.index,
    y=conversion_data.values
)

# Add chart title
plt.title("Average Conversion Rate by Traffic Source")

# Add axis labels
plt.xlabel("Traffic Source")

plt.ylabel("Average Conversion Rate")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 16: CORRELATION HEATMAP
# ==========================================

# Select numerical columns
numeric_data = cleaned_web.select_dtypes(
    include="number"
)

# Calculate correlation
correlation = numeric_data.corr()

# Set figure size
plt.figure(figsize=(10, 7))

# Create heatmap
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

# Add chart title
plt.title("Website Traffic Correlation Heatmap")

# Display chart
plt.show()
```

```python

# ==========================================
# STEP 17: IDENTIFY BEST TRAFFIC SOURCE
# ==========================================

# Find highest conversion rate
best_source = conversion_data.idxmax()

# Find highest conversion value
best_conversion = conversion_data.max()

# Display result
print("Best Performing Traffic Source:")

print("Traffic Source:", best_source)

print("Average Conversion Rate:", round(best_conversion, 3))
```
